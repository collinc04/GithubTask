# Project 2 - Fall with Air Drag - Ariana Rider

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------- Constants -------------------------------------------
yo = 200
vo = 0
dt = 0.001
g = 9.81
r = 0.115
C = 0.25
p = 1.2
m = 0.425
A = np.pi*r**2
t = 0
y = yo
v = vo
a = -g

vfin = -np.sqrt((2*m*g) / (C*A*p))

# ------------------------------ Lists ----------------------------------
t_ = []
y_ = []
v_ = []
a_ = []

def append(t, y, v, a):
    t_.append(t)
    y_.append(y)
    v_.append(v)
    a_.append(a)


# ------------------------------- Motion -------------------------------------------
while y > 0:
    append(t, y, v, a)
    
    drag = -0.5 * p * C * A * v*abs(v)
    fnet = drag - m*g
    a = fnet/m
    #ball.pos = ball.pos + v*dt

    #if ball.pos.x > wall.pos.x - ball.radius - wall.size.x/2:
        #v.x = -v.x
        #ball.pos.x = wall.pos.x - ball.radius - wall.size.x/2
        
    
    v = v + a * dt
    y = y + v * dt
    t = t+dt

# ------------------------------- Display Graphs -------------------------------------------
# Position vs time
plt.figure()
plt.plot(t_, y_)
plt.xlabel("Time (s)")
plt.ylabel("Height (m)")
plt.title("Position vs Time")
plt.show()

# Velocity vs time
plt.figure()
plt.plot(t_, v_)
plt.axhline(vfin, linestyle='--')   # terminal velocity line
plt.xlabel("Time (s)")
plt.ylabel("Velocity (m/s)")
plt.title("Velocity vs Time")
plt.show()

# Acceleration vs time
plt.figure()
plt.plot(t_, a_)
plt.xlabel("Time (s)")
plt.ylabel("Acceleration (m/s^2)")
plt.title("Acceleration vs Time")
plt.show()
```
