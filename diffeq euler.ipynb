```python
t = 0 # initial t-value
y = 1 # initial y-value
h = 0.1 # step size

while t <= 5:
    if t == 0 or round(t, 1) == 1 or round(t, 1) == 2 or round(t, 1) == 3 or round(t, 1) == 4 or round(t, 1) == 5:
        print(round(t),y)
    f = 3-2*t-0.5*y # differential equation
    y = y + f*h # euler method
    t = t + h
```
